# 00 — Pull Raw Data

Load and validate all raw source files. **No transformations here** — only shape checks
and column previews to confirm files loaded correctly before any downstream work.

| | |
|---|---|
| **Inputs** | `data/raw/renewables.xlsx`, `Prod_dataset.xlsx`, `spectrum.csv`, `plant.xlsx` |
| **Outputs** | None (validation only) |

```python
from pathlib import Path
import pandas as pd
import numpy as np

# ── Paths (all relative — never hardcode absolute paths) ──────────────────────
DATA_RAW = Path("../data/raw")
```

```python
# ── Load survey data ──────────────────────────────────────────────────────────
survey = pd.read_excel(DATA_RAW / "renewables.xlsx")
print(f"Survey shape: {survey.shape}")
print(f"Columns: {list(survey.columns)}")
survey.head(3)
```

```python
# ── Check dependent variable: Q137 (RE support 1–5) ──────────────────────────
print("Q137 value counts:")
print(survey["Q137"].value_counts().sort_index())
print(f"\nMean: {survey['Q137'].mean():.3f}  |  SD: {survey['Q137'].std():.3f}")
print(f"Missing: {survey['Q137'].isna().sum()}")
```

```python
# ── Load EIA SEDS coal production data ───────────────────────────────────────
coal = pd.read_excel(DATA_RAW / "Prod_dataset.xlsx")
print(f"Coal production shape: {coal.shape}")
print(f"Columns: {list(coal.columns)}")
coal.head(3)
```

```python
# ── Check year range in coal data ─────────────────────────────────────────────
# Adjust column name below if your year column is named differently
year_col = [c for c in coal.columns if "year" in c.lower() or "Year" in c]
if year_col:
    print(f"Year range: {coal[year_col[0]].min()} – {coal[year_col[0]].max()}")
else:
    print("Could not auto-detect year column. Check columns above.")
```

```python
# ── Load demographics file ────────────────────────────────────────────────────
demo = pd.read_csv(DATA_RAW / "spectrum.csv")
print(f"Demographics shape: {demo.shape}")
print(f"Columns: {list(demo.columns)}")
demo.head(3)
```

```python
# ── Load power plant locations ────────────────────────────────────────────────
plants = pd.read_excel(DATA_RAW / "plant.xlsx")
print(f"Plant data shape: {plants.shape}")
print(f"Columns: {list(plants.columns)}")
plants.head(3)
```

```python
# ── Summary validation ────────────────────────────────────────────────────────
print("=" * 50)
print("VALIDATION SUMMARY")
print("=" * 50)
for name, df in [("survey", survey), ("coal", coal), ("demo", demo), ("plants", plants)]:
    missing = df.isna().sum().sum()
    print(f"{name:10s}  rows={df.shape[0]:6d}  cols={df.shape[1]:3d}  total_missing={missing}")
print("\nAll files loaded successfully. Proceed to 01_merge.ipynb.")
```
